# Group15 Express Coffee — Project Notebook

Notebook ต้นฉบับ `Sourcecode.ipynb` เป็น CLI รุ่นแรกแบบไฟล์เดียว ความสามารถหลักได้พัฒนาต่อในโปรเจกต์ปัจจุบันแล้ว โดยแยกตามหน้าที่ดังนี้:

| ต้นแบบใน Notebook | โค้ดปัจจุบัน |
|---|---|
| `MENU`, `MEMBERS` | ข้อมูลเริ่มต้นใน `coffee_pos/services/shop_service.py` และ JSON ใน `data/` |
| `get_customization()`, `take_order()`, `checkout()` | `coffee_pos/ui/cli.py`, `coffee_pos/services/shop_service.py` และ `coffee_pos/models/` |
| `show_daily_summary()` | สรุปยอดใน `ShopService` และ Sales Advisor |
| `main()` | entry point ที่ `main.py` |

รันเซลล์ถัดไปเพื่อเปิดแอปเวอร์ชันปัจจุบัน ซึ่งเก็บข้อมูลลง JSON และมีเมนูจัดการเพิ่มเติม ใช้ Python environment ที่ติดตั้ง `requirements.txt` แล้ว และเปิด Notebook จากโฟลเดอร์โปรเจกต์นี้

```python
from pathlib import Path
import sys

project_root = next(
    (path.resolve() for path in (Path.cwd(), *Path.cwd().parents)
     if (path / 'main.py').is_file()),
    None,
)
if project_root is None:
    raise RuntimeError('เปิด Notebook จากภายในโฟลเดอร์โปรเจกต์ final-project')

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from main import main
main()
```
